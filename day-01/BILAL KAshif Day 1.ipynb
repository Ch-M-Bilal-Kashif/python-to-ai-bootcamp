```python
#marks analyzer also user input based

def calculate_total(marks):
    return sum(marks)

def calculate_average(marks):
    return sum(marks) / len(marks)

def highest_marks(marks):
    return max(marks)

def lowest_marks(marks):
    return min(marks)

def count_pass(marks):
    count = 0

    for mark in marks:
        if mark >= 50:
            count = count + 1

    return count

name = input("Enter student name: ")

marks = []

for i in range(5):
    mark = int(input("Enter marks: "))
    marks.append(mark)

print("Student:", name)
print("Marks:", marks)
print("Total:", calculate_total(marks))
print("Average:", calculate_average(marks))
print("Highest:", highest_marks(marks))
print("Lowest:", lowest_marks(marks))
print("Passed Subjects:", count_pass(marks))
print("Failed Subjects:", len(marks) - count_pass(marks))
```
