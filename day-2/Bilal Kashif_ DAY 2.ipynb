```python
#custom questions

print("Q1: did passenger class affect survival rates?")


first = df.loc[df["Pclass"] == 1, "Survived"]
second = df.loc[df["Pclass"] == 2, "Survived"]
third = df.loc[df["Pclass"] == 3, "Survived"]

print("First Class Survival Rate:", first.mean() * 100)
print("Second Class Survival Rate:", second.mean() * 100)
print("hird Class Survival Rate:", third.mean() * 100)


print("/n")




#multiplying with one hundred gives the proper percentage


print("Q2: Which age group had the highest survival rate on the Titanic?")

children = df.loc[df["Age"] < 18, "Survived"]
adults = df.loc[(df["Age"] >= 18) & (df["Age"] < 60), "Survived"]
seniors = df.loc[df["Age"] >= 60, "Survived"]

print("Children Survival Rate:", children.mean() * 100)
print("Adults Survival Rate:", adults.mean() * 100)
print("Seniors Survival Rate:", seniors.mean() * 100)
```
